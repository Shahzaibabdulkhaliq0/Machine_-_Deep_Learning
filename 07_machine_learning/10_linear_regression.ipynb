**Linear Regression = A Machine Learning algorithm used to predict continuous numerical values.**

It learns the relationship between **input (`X`)** and **target (`y`)** to make predictions.

### 📚 Types

* 1️⃣ **Simple Linear Regression** → One feature
* 2️⃣ **Multiple Linear Regression** → Multiple features
* 3️⃣ **Polynomial Regression** → Curved relationship

### ❓ Why Use It?

* 🎯 Predict continuous values
* 🔗 Understand relationships between variables
* 📊 Simple and easy to interpret

### 📍 Where Is It Used?

* 🏠 House Price Prediction
* 💰 Sales & Revenue Prediction
* 📦 Demand Prediction
* 🌡️ Temperature Prediction
* 💼 Salary Prediction

### 🌍 Simple Real-World Example

🏠 **House Price Prediction**

`House Size → Linear Regression → Predicted House Price`

For example, the model learns from previous houses:

**Size ↑ → Price usually ↑**

Then it can predict the price of a new house based on its size.

### ✅ Main Assumptions

* 📏 **Linearity** → Relationship should be roughly linear
* 🔗 **Independence** → Observations should be independent
* 📊 **Constant Variance** → Error spread should be roughly constant
* 🔔 **Normal Residuals** → Errors should be approximately normal
* 🚫 **No Perfect Multicollinearity** → Features should not be perfectly related

> 💡 **Rule:** Linear Regression → **Linear relationship → Continuous-value prediction.**
---


```python
# Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score
```

```python
# Load Dataset
df = sns.load_dataset('tips')
df.head()
```

```python
# (Optional) Normally Check Datapoints
sns.scatterplot(x='total_bill', y='tip', data=df)
```

```python
# Split the Data into X & y
X = df[['total_bill']]       ## Input Value
y = df['tip']                ## Output Value
```

```python
# Split Data into Train & Test
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

```python
# Call the model
model = LinearRegression()
```

```python
# Train the Model
model.fit(X_train, y_train)
```

```python
# Predict
y_pred = model.predict(X_test)
```

```python
# Evaluate the Model
print('MSE = ', mean_squared_error(y_test, y_pred))
print('R2 = ', r2_score(y_test, y_pred))
print('RMSE = ', np.sqrt(mean_squared_error(y_test, y_pred)))
```

```python
# Plot the Model to Check Best fit Line
plt.scatter(X_test, y_test)
plt.plot(X_test, y_pred, color='red')
plt.show()
```

### ⚖️ What About Scaling?

* 🚫 **We did not use a Scaler in this notebook.**
* 📏 Scalers can be applied to **numerical input features (`X`)** when needed.
* 🔢 Common choices: **StandardScaler, MinMaxScaler, RobustScaler**.
* ⚠️ Scaling is **not mandatory for every Linear Regression problem**.
* 💡 If features have very different scales, scaling can make the model easier to work with.

### 🔤 What About Encoding?

* 🚫 **We did not use an Encoder in this notebook.**
* 🔤 Encoders are needed when `X` contains **categorical/text features** such as `Male/Female` or `City`.
* 🔢 Common choices: **One-Hot, Ordinal, Binary Encoding**.
* 🎯 The **target (`y`) in Linear Regression is normally continuous numerical data**, so categorical encoding is not used to turn the regression target into categories.

> 💡 **Final Rule:**
> **Continuous `y` → Regression**
> **Categorical `X` → Encode if needed**
> **Numerical `X` → Scale if needed**
> **Choose preprocessing based on your data and model.**
