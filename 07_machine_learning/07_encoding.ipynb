**Feature Encoding = Converting categorical/text data into numerical values so Machine Learning models can process it.**

### ❓ Why Do We Need Encoding?

- 🤖 ML models mainly work with **numerical data**.
- 🔤 Categorical values like `Male`, `Female`, `Pakistan`, `India` are **text**, not numbers.
- 🔢 Encoding converts these categories into **numbers**.
- 📊 This makes categorical features usable for **training and prediction**.

### Example

Before:

`Gender = Male, Female, Male`

After Encoding:

`Gender = 1, 0, 1`

> 💡 **Rule:** **Categorical Data → Numerical Data → ML Model**
---

<h3 style="color:orange; font-weight:bold">🏷️ Label Encoding</h3>

- **Label Encoding = Converts each category into a numerical label.**
- Each unique category gets a **different integer**.
- ✅ Useful for **ordinal categories** where order matters.

### Example

`Low → 0`  
`Medium → 1`  
`High → 2`

👉 The numbers represent the **order of the categories**.

> 💡 **Rule:** Label Encoding → **Categories → Integer Labels**
---

<h3 style="color:orange; font-weight:bold">🏷️ OrdinalEncoder</h3>

- **OrdinalEncoder = Converts categorical values into numerical values based on their order.**
- ✅ Best for **ordinal categories** where a meaningful ranking exists.
- You can **define the order** of the categories.

### Example

`Low → 0`  
`Medium → 1`  
`High → 2`

👉 The numbers represent the **real order** of the categories.

> 💡 **Rule:** OrdinalEncoder → **Ordered Categories → Ordered Numbers**

```python
# Ordinal Encoding 'day' Column Using Specific Order
encode = OrdinalEncoder(categories=[['Thur', 'Fri', 'Sat', 'Sun']])
```

```python
# Run Model for Convert into Ordinal Encoding
df['encoded_day'] = encode.fit_transform(df[['day']])
```

```python
df.head()
```

<h3 style="color:orange; font-weight:bold">🔢 One-Hot Encoding</h3>

- **One-Hot Encoding = Converts each category into a separate binary column.**
- Each category gets **`1` if present and `0` otherwise**.
- ✅ Best for **nominal categories** where there is **no meaningful order**.

### Example

`Color = Red, Blue, Green`

After One-Hot Encoding:

| Color | Red | Blue | Green |
|---|---:|---:|---:|
| Red | 1 | 0 | 0 |
| Blue | 0 | 1 | 0 |
| Green | 0 | 0 | 1 |

> 💡 **Rule:** One-Hot Encoding → **One Category = One Binary Column.**

```python
# One-Hot Encoding on 'day' Column
encoder = OneHotEncoder()
```

```python
# Run Model for Convert into One-Hot Encoding
encode = encoder.fit_transform(df[['sex']]).toarray()
```

```python
# Convert into Pandas DataFrame
data = pd.DataFrame(encode)
```

<h3 style="color:orange; font-weight:bold">🔢 Binary Encoding</h3>

- **Binary Encoding = Converts categories into binary (0/1) values using binary digits.**
- It uses **fewer columns** than One-Hot Encoding when there are many categories.
- ✅ Useful for **high-cardinality categorical features**.

### Example

Categories:

`Red, Blue, Green, Yellow`

Binary codes:

`Red → 00`  
`Blue → 01`  
`Green → 10`  
`Yellow → 11`

> 💡 **Rule:** Binary Encoding → **Categories → Binary digits → Fewer columns.**

## 🎯 Feature Encoding — Final Summary

Feature Encoding converts **categorical data into numerical form** so that Machine Learning models can process it.

### 🔑 Methods Covered

- 🏷️ **Label Encoding** → Categories → Integer labels
- 📊 **Ordinal Encoding** → Ordered categories → Ordered numbers
- 🔢 **One-Hot Encoding** → Categories → Binary columns
- 💻 **Binary Encoding** → Categories → Binary digits

> 💡 **Final Rule:** Choose the encoding method based on whether your categories have a **meaningful order** and how many categories your feature contains.
